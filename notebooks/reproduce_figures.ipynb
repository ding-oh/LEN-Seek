# Reproducing the figures and tables

This notebook computes the figures and tables of the manuscript from intermediate results, one result per cell, through `analysis/run.py` (the same commands as `examples/analysis.sh`). A cell either runs its result or reports the inputs it is missing; the last cell lists what ran, what was skipped and what failed.

**Inputs.** Set the environment variables listed by `python analysis/run.py -h` before starting Jupyter (`LENSEEK_EVAL_DIR`, `LENSEEK_PAIRS`, `LENSEEK_SCAN_DIR`, `LENSEEK_REFERENCE_DIR`, ...). Outputs go to `$LENSEEK_OUT` (default `out/` in the repository). The intermediate files that the results read are written by the scripts in `scripts/`: the site annotation, the per-site checks of the BsitePDB files and the top-100 lists of the sequence-identity and random baselines by `examples/analysis.sh inputs`, the timing runs by `examples/analysis.sh timing`, the structural labels, the binding sites from later PDB entries and the larger-database study by `examples/labels.sh`, `examples/holdout.sh` and `examples/scaling.sh`. Figures that are drawings or molecular renderings are not made here.

```python
import os
import subprocess
import sys
import time

REPO = os.path.abspath('..') if os.path.basename(os.getcwd()) == 'notebooks' else os.getcwd()
os.environ.setdefault('LENSEEK_OUT', os.path.join(REPO, 'out'))
STATUS = []


def result(name, *extra):
    """Run one result of analysis/run.py and record whether it ran, was skipped (missing inputs) or failed."""
    t0 = time.time()
    p = subprocess.run([sys.executable, os.path.join(REPO, 'analysis', 'run.py'), name, *extra], cwd=REPO,
                       capture_output=True, text=True)
    print((p.stdout + p.stderr)[-3000:])
    status = {0: 'ran', 3: 'skipped'}.get(p.returncode, 'failed')
    skip = [l[len('[skip] '):] for l in p.stdout.splitlines() if l.startswith('[skip]')]
    STATUS.append((name, status, skip[0].split(': ', 1)[1] if skip else f'{time.time() - t0:.0f} s'))
```
